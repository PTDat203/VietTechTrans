# 05.02 — Build General Test

Build hoặc kiểm tra General Test độc lập FLORES-200 `devtest` (EN↔VI). Script sẽ từ chối build nếu phát hiện overlap với train, validation hoặc IT Test.

```python
from pathlib import Path
import json
import subprocess
import sys

ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
elif ROOT.name == '05_dataset_building':
    ROOT = ROOT.parents[1]
ROOT
```

```python
CONFIG = ROOT / 'configs/data/general_test_flores200_devtest_v1.json'
config = json.loads(CONFIG.read_text(encoding='utf-8'))
config
```

```python
GENERAL_DIR = ROOT / 'data/processed/general_test_flores200_devtest_v1'
manifest_path = GENERAL_DIR / 'dataset_manifest.json'
leakage_path = GENERAL_DIR / 'leakage_report.json'

if manifest_path.exists():
    print(json.dumps(json.loads(manifest_path.read_text(encoding='utf-8')), ensure_ascii=False, indent=2))
else:
    print('General Test chưa được build.')

if leakage_path.exists():
    print('\nLeakage report:')
    print(json.dumps(json.loads(leakage_path.read_text(encoding='utf-8')), ensure_ascii=False, indent=2))
```

## Rebuild có chủ ý

Chỉ đặt `REBUILD = True` khi bạn muốn tạo lại release từ đúng hai file FLORES đã tải. Lệnh mặc định không ghi đè release hiện có.

```python
REBUILD = False
EN_FILE = ROOT / 'data/raw/flores200/flores200_dataset/devtest/eng_Latn.devtest'
VI_FILE = ROOT / 'data/raw/flores200/flores200_dataset/devtest/vie_Latn.devtest'

if REBUILD:
    command = [sys.executable, str(ROOT / 'tools/build_general_test.py'), '--en', str(EN_FILE), '--vi', str(VI_FILE), '--force']
    subprocess.run(command, cwd=ROOT, check=True)
else:
    print('REBUILD=False — không thay đổi General Test.')
```
